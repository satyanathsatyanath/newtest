```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .remote("sc://localhost") \
    .getOrCreate()
```

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.remote("sc://localhost:15002").getOrCreate()
spark.sql("SELECT 'Connected' as status").show()
```

```python
from pyspark.sql import SparkSession

# Use the IP address instead of localhost
spark = SparkSession.builder \
    .remote("sc://127.0.0.1:15002") \
    .getOrCreate()

spark.sql("SELECT 'Instant Connection' as status").show()
```
